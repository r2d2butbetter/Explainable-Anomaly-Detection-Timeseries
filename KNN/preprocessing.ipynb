# Spambase Data Preprocessing

This notebook loads the raw spambase.data file, cleans it, normalizes features, and saves the processed dataset.

```python
import pandas as pd
import numpy as np

# Proper feature names for spambase dataset
feature_names = [
    'word_freq_make', 'word_freq_address', 'word_freq_all', 'word_freq_3d', 'word_freq_our', 
    'word_freq_over', 'word_freq_remove', 'word_freq_internet', 'word_freq_order', 'word_freq_mail', 
    'word_freq_receive', 'word_freq_will', 'word_freq_people', 'word_freq_report', 'word_freq_addresses', 
    'word_freq_free', 'word_freq_business', 'word_freq_email', 'word_freq_you', 'word_freq_credit', 
    'word_freq_your', 'word_freq_font', 'word_freq_000', 'word_freq_money', 'word_freq_hp', 
    'word_freq_hpl', 'word_freq_george', 'word_freq_650', 'word_freq_lab', 'word_freq_labs', 
    'word_freq_telnet', 'word_freq_857', 'word_freq_data', 'word_freq_415', 'word_freq_85', 
    'word_freq_technology', 'word_freq_1999', 'word_freq_parts', 'word_freq_pm', 'word_freq_direct', 
    'word_freq_cs', 'word_freq_meeting', 'word_freq_original', 'word_freq_project', 'word_freq_re', 
    'word_freq_edu', 'word_freq_table', 'word_freq_conference', 'char_freq_;', 'char_freq_(', 
    'char_freq_[', 'char_freq_!', 'char_freq_$', 'char_freq_#', 'capital_run_length_average', 
    'capital_run_length_longest', 'capital_run_length_total', 'label'
]

# Load the dataset with proper column names
data = pd.read_csv('spambase.data', header=None, names=feature_names)

print(f"Dataset shape: {data.shape}")
print(f"Feature names (first 10): {list(data.columns[:10])}")
print(f"\nLabel distribution:\n{data['label'].value_counts()}")
```

```python
# Check for missing values
if data.isnull().sum().sum() > 0:
    print("Warning: Missing values found. Filling with mean.")
    data = data.fillna(data.mean(numeric_only=True))
else:
    print("No missing values detected.")
```

```python
# Normalize features to [0,1] using min-max scaling
X = data.iloc[:, :-1].values
y = data.iloc[:, -1].values

X_normalized = (X - X.min(axis=0)) / (X.max(axis=0) - X.min(axis=0) + 1e-8)

print(f"Normalization complete. Feature range: [{X_normalized.min():.4f}, {X_normalized.max():.4f}]")
```

```python
# Shuffle the data for randomness
np.random.seed(42)
indices = np.random.permutation(len(X_normalized))
X_shuffled = X_normalized[indices]
y_shuffled = y[indices]

# Save processed data with meaningful feature names
processed_data = pd.DataFrame(X_shuffled, columns=feature_names[:-1])  # All features except 'label'
processed_data['label'] = y_shuffled

processed_data.to_csv('spambase_processed_data.csv', index=False)

print("Processed data saved with meaningful feature names!")
print(f"Saved columns: {list(processed_data.columns[:10])}... (showing first 10)")
```
